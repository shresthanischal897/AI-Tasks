```python
import numpy as np
import pandas as pd
```

```python
def logistic_function(x):
    return 1 / (1 + np.exp(-x))

# Test the function
print("Sigmoid of 0:", logistic_function(0))  # 0.5
print("Sigmoid of [0,2,-3]:", logistic_function(np.array([0,2,-3])))  # [0.5,0.881,0.047]
```

```python
def log_loss(y_true, y_pred):
    y_pred = np.clip(y_pred, 1e-10, 1-1e-10)  # avoid log(0)
    return -(y_true*np.log(y_pred) + (1-y_true)*np.log(1-y_pred))

# Test
print("Log loss for (y=1, y_pred=0.8):", log_loss(1, 0.8))
print("Log loss for (y=0, y_pred=0.2):", log_loss(0, 0.2))
```

```python
def cost_function(y_true, y_pred):
    n = len(y_true)
    return np.mean(log_loss(y_true, y_pred))

# Test with multiple samples
y_true = np.array([1, 0, 1])
y_pred = np.array([0.9, 0.1, 0.8])
print("Cost:", cost_function(y_true, y_pred))
```

```python
def costfunction_logreg(X, y, w, b):
    z = np.dot(X, w) + b
    y_pred = logistic_function(z)
    return cost_function(y, y_pred)
```

```python
def compute_gradient(X, y, w, b):
    n = X.shape[0]
    z = np.dot(X, w) + b
    y_pred = logistic_function(z)
    grad_w = -(1/n) * np.dot((y - y_pred), X)
    grad_b = -(1/n) * np.sum(y - y_pred)
    return grad_w, grad_b
```

```python
def gradient_descent(X, y, w, b, alpha, n_iter):
    cost_history = []
    for i in range(n_iter):
        grad_w, grad_b = compute_gradient(X, y, w, b)
        w -= alpha * grad_w
        b -= alpha * grad_b
        cost = costfunction_logreg(X, y, w, b)
        cost_history.append(cost)
    return w, b, cost_history
```

```python
def prediction(X, w, b, threshold=0.5):
    y_prob = logistic_function(np.dot(X, w) + b)
    return (y_prob >= threshold).astype(int)
```

```python
def evaluate_classification(y_true, y_pred):
    TP = np.sum((y_true==1) & (y_pred==1))
    TN = np.sum((y_true==0) & (y_pred==0))
    FP = np.sum((y_true==0) & (y_pred==1))
    FN = np.sum((y_true==1) & (y_pred==0))

    precision = TP / (TP + FP) if (TP + FP) > 0 else 0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0
    f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    return {
        "confusion_matrix": np.array([[TN, FP],[FN, TP]]),
        "precision": precision,
        "recall": recall,
        "f1_score": f1_score
    }
```

```python
# Sample dataset
X = np.array([[0.1, 0.2], [-0.1, 0.1]])
y = np.array([1, 0])

# Initialize parameters
w = np.zeros(X.shape[1])
b = 0.0
alpha = 0.1
n_iter = 1000

# Train the model
w_out, b_out, cost_history = gradient_descent(X, y, w, b, alpha, n_iter)

# Make predictions
y_pred = prediction(X, w_out, b_out)

# Evaluate
metrics = evaluate_classification(y, y_pred)

# Print outputs
print("\nFinal weights:", w_out)
print("Bias:", b_out)
print("First 10 cost values:", cost_history[:10])
print("Predictions:", y_pred)
print("Metrics:", metrics)
```
