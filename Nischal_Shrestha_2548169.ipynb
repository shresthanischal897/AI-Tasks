```python
'''6. Adding HDI Category Column:
• Create a new column HDI Category that classifies each country into one of the four official
Human Development Index groups. The classification should be based on the HDI value for the
year 2022. Use the following categories and thresholds defined by the United Nations Development
Programme (UNDP):
HDI Category HDI Range (hdi)
Low < 0.550
Medium 0.550 – 0.699
High 0.700 – 0.799
Very ≥ 0.800
After creating this new column:
• verify that all countries are classified correctly,
• ensure the updated dataframe includes the new category column.
• Save the final dataframe as HDI category added.csv and include this file in your final
submission.
'''
def hdi_cat(hdi):
    if hdi < 0.550:
        return 'Low'
    elif hdi <= 0.699:
        return 'Medium'
    elif hdi <= 0.799:
        return 'High'
    else:
        return 'Very High'


df22['HDI_Category'] = df22['hdi'].apply(hdi_cat)

df22[['country', 'hdi', 'HDI_Category']].head(10)

df22['HDI_Category'].value_counts()

df22.to_csv('HDI_category_added.csv', index=False)
```

```python
'''Outlier Detection:
• Detect outliers in HDI and GNI per Capita using the 1.5 × IQR rule.'''
```

```python
sa_2022 = df_sa[df_sa['year'] == 2022]

q1_hdi = sa_2022['hdi'].quantile(0.25)
q3_hdi = sa_2022['hdi'].quantile(0.75)
iqr_hdi = q3_hdi - q1_hdi

hdi_out = sa_2022[
    (sa_2022['hdi'] < q1_hdi - 1.5*iqr_hdi) |
    (sa_2022['hdi'] > q3_hdi + 1.5*iqr_hdi)
]

q1_gni = sa_2022['gross_inc_percap'].quantile(0.25)
q3_gni = sa_2022['gross_inc_percap'].quantile(0.75)
iqr_gni = q3_gni - q1_gni

gni_out = sa_2022[
    (sa_2022['gross_inc_percap'] < q1_gni - 1.5*iqr_gni) |
    (sa_2022['gross_inc_percap'] > q3_gni + 1.5*iqr_gni)
]

print('HDI Outliers:')
print(hdi_out[['country','hdi']])

print('\nGNI Outliers:')
print(gni_out[['country','gross_inc_percap']])
```

```python
'''6. Correlation Analysis:
• For each region, compute correlations of HDI with:
– Gender Development Index
– Life Expectancy Index
• Create scatter plots with trendlines for each correlation.
• Interpret the strength and direction of these relationships.'''
```

```python
import numpy as np

def pearson_corr(x, y):
    xm, ym = np.mean(x), np.mean(y)
    num = np.sum((x - xm)*(y - ym))
    den = np.sqrt(np.sum((x - xm)**2) * np.sum((y - ym)**2))
    return num / den

print("---- South Asia Correlations ----")
for m in metrics:
    r = pearson_corr(sa['hdi'], sa[m])
    print(m, "vs HDI:", round(r,4))

print("\n---- Middle East Correlations ----")
for m in metrics:
    r = pearson_corr(me['hdi'], me[m])
    print(m, "vs HDI:", round(r,4))
```

```python
def scatter_with_trend(df, x, y, title):
    plt.figure(figsize=(6,4))
    plt.scatter(df[x], df[y], color='blue')

    m, b = np.polyfit(df[x], df[y], 1)
    plt.plot(df[x], m*df[x] + b, color='red', linewidth=2)

    plt.title(title)
    plt.xlabel(x)
    plt.ylabel(y)
    plt.grid(True)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

scatter_with_trend(sa, 'gender_development', 'hdi',
                   'South Asia: HDI vs Gender Development Index')

scatter_with_trend(sa, 'life_expectancy', 'hdi',
                   'South Asia: HDI vs Life Expectancy Index')

scatter_with_trend(me, 'gender_development', 'hdi',
                   'Middle East: HDI vs Gender Development Index')

scatter_with_trend(me, 'life_expectancy', 'hdi',
                   'Middle East: HDI vs Life Expectancy Index')
```

```python
'''Outlier Detection:
• Detect outliers in HDI {"hdi"} and GNI per Capita {"gross inc percap"} for each region using
the 1.5 × IQR rule.
• Create scatter plots highlighting outliers in a different color.
• Discuss the significance of these outliers.'''
```

```python
import pandas as pd
import matplotlib.pyplot as plt

sa = pd.read_csv('HDI_SouthAsia_2020_2022.csv')
me = pd.read_csv('HDI_MiddleEast_2020_2022.csv')

for c in ['hdi','gross_inc_percap']:
    sa[c] = pd.to_numeric(sa[c], errors='coerce')
    me[c] = pd.to_numeric(me[c], errors='coerce')

def detect_outliers(df, col):
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5*iqr
    upper = q3 + 1.5*iqr
    return df[(df[col] < lower) | (df[col] > upper)]

sa_hdi_out = detect_outliers(sa, 'hdi')
sa_gni_out = detect_outliers(sa, 'gross_inc_percap')

me_hdi_out = detect_outliers(me, 'hdi')
me_gni_out = detect_outliers(me, 'gross_inc_percap')

print("South Asia HDI Outliers:\n", sa_hdi_out[['country','hdi']])
print("South Asia GNI Outliers:\n", sa_gni_out[['country','gross_inc_percap']])

print("\nMiddle East HDI Outliers:\n", me_hdi_out[['country','hdi']])
print("Middle East GNI Outliers:\n", me_gni_out[['country','gross_inc_percap']])
```

```python
def plot_outliers(df, x, y, title):
    plt.figure(figsize=(6,4))

    q1 = df[y].quantile(0.25)
    q3 = df[y].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5*iqr
    upper = q3 + 1.5*iqr

    out = df[(df[y] < lower) | (df[y] > upper)]
    normal = df[(df[y] >= lower) & (df[y] <= upper)]

    plt.scatter(normal[x], normal[y], color='blue', label='Normal')
    plt.scatter(out[x], out[y], color='red', label='Outlier')

    plt.title(title)
    plt.xlabel(x)
    plt.ylabel(y)
    plt.legend()
    plt.grid(True)
    plt.show()


# South Asia
plot_outliers(sa, 'country', 'hdi', 'South Asia: HDI Outliers')
plot_outliers(sa, 'country', 'gross_inc_percap', 'South Asia: GNI per Capita Outliers')

# Middle East
plot_outliers(me, 'country', 'hdi', 'Middle East: HDI Outliers')
plot_outliers(me, 'country', 'gross_inc_percap', 'Middle East: GNI per Capita Outliers')
```

```python
'''The South Asian region also had no outliers of HDI or GNI, which implies that development is rather homogeneous.
Yemen appears to be a significant exception to the Middle East, and it demonstrates the significant issues with human development that Yemen can only contend with.
The significance of outliers is that they identify extremities and help in showing disparities between the regions and assist the policymakers to concentrate on the interventions that are required by the countries, which are way out of the regional average.'''
```
